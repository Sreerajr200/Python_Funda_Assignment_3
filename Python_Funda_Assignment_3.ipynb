Q1) ******************* Number Guessing Game using While Loop ******************* 

```python
import random
secret_number  = random.randint(1,10)
attempts = 3
i = 1

print("Attempt ",i,":")
user_number  = int(input("Please guess the number from 1 to 10: "))

while i <= attempts-1:
    if user_number < 1 or user_number > 10:
        i = i+1
        print("Attempt ",i,":")
        user_number  = int(input("Please guess a number within the range 1 to 10."))
        continue
    elif user_number < secret_number:
        i = i+1
        print("Attempt ",i,":")
        user_number  = int(input("The guess is lower than the secret number. Please try again :"))
    elif user_number > secret_number:
        i = i+1
        print("Attempt ",i,":")
        user_number  = int(input("The guess is higher than the secret number. Please try again :"))
    else:
        print("Congratulations!!!! Your guess is correct!")
        break

if i >= attempts or user_number != secret_number:
    print("Secret number was ",secret_number)
    print("Better luck next time. Game over!!!")
```

Q2) ******************* Multiplication Table ******************* 

```python
user_number = int(input("Please eneter a number:"))
print("The multiplication table of",user_number,"is below:")

for i in range(1,11):
    print(user_number,"X",i,"=",user_number*i)
```

Q3)******************* BMI Calculator *******************  

```python
def calculate_bmi(weight, height):
    return round(weight/height**2,2)
    
w = float(input("Please eneter your weight in kg:"))
h = float(input("Please eneter your height in meter:"))
print("Your Body Mass Index is : ",calculate_bmi(w,h))
```
